```python
import tensorflow as tf
from tensorflow.keras.layers import *
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.applications import EfficientNetB0
import tensorflow.keras.backend as K
import numpy as np
import cv2
import glob
import os
from tqdm import tqdm
import random

print("="*80)
print("AI SPILLGUARD - COMPLETE CORRECTED TRAINING & SAVING PIPELINE")
print("="*80)

# ============================================================================
# CRITICAL STEP 0: REGISTER ALL CUSTOM FUNCTIONS FIRST (BEFORE EVERYTHING)
# ============================================================================
print("\n[STEP 0] Registering all custom loss and metric functions...")

# Register Loss Function
@tf.keras.utils.register_keras_serializable(package="SpillGuard", name="optimized_combined_loss")
def optimized_combined_loss(y_true, y_pred):
    """
    Combined loss: Weighted BCE + Dice + Focal
    MUST be registered BEFORE model compilation
    """
    # Weighted binary crossentropy
    pos_weight = tf.reduce_sum(1. - y_true) / (tf.reduce_sum(y_true) + K.epsilon())
    pos_weight = tf.clip_by_value(pos_weight, 1.0, 100.0)
    bce = K.binary_crossentropy(y_true, y_pred)
    wbce = tf.where(K.equal(y_true, 1), pos_weight * bce, bce)
    wbce_loss = K.mean(wbce)

    # Dice loss
    dice = 1 - (2 * K.sum(y_true * y_pred) + 1) / (K.sum(y_true) + K.sum(y_pred) + 1)

    # Focal loss
    epsilon = K.epsilon()
    y_pred_clipped = K.clip(y_pred, epsilon, 1.0 - epsilon)
    p_t = tf.where(K.equal(y_true, 1), y_pred_clipped, 1 - y_pred_clipped)
    focal = -K.log(p_t) * K.pow(1 - p_t, 3.0)
    focal_loss_val = K.mean(focal)

    # Weighted combination
    return 0.3 * wbce_loss + 0.5 * dice + 0.2 * focal_loss_val

# Register Metric Functions
@tf.keras.utils.register_keras_serializable(package="SpillGuard", name="pixel_accuracy")
def pixel_accuracy(y_true, y_pred, threshold=0.5):
    """Pixel-level accuracy metric"""
    y_true_bin = K.round(y_true)
    y_pred_bin = K.cast(y_pred > threshold, 'float32')
    correct = K.cast(K.equal(y_true_bin, y_pred_bin), 'float32')
    return K.mean(correct)

@tf.keras.utils.register_keras_serializable(package="SpillGuard", name="dice_coefficient")
def dice_coefficient(y_true, y_pred, smooth=1e-6):
    """Dice coefficient metric"""
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

@tf.keras.utils.register_keras_serializable(package="SpillGuard", name="iou_score")
def iou_score(y_true, y_pred, threshold=0.5):
    """IoU (Intersection over Union) metric"""
    y_pred_bin = tf.cast(y_pred > threshold, tf.float32)
    inter = tf.reduce_sum(y_true * y_pred_bin)
    union = tf.reduce_sum(y_true) + tf.reduce_sum(y_pred_bin) - inter
    return (inter + K.epsilon()) / (union + K.epsilon())

print("✅ Custom loss registered: optimized_combined_loss")
print("✅ Custom metrics registered: pixel_accuracy, dice_coefficient, iou_score")
print(f"   TensorFlow version: {tf.__version__}")

# ============================================================================
# STEP 1: Reproducibility
# ============================================================================
print("\n[STEP 1] Setting random seeds for reproducibility...")

def set_seeds(seed=42):
    np.random.seed(seed)
    random.seed(seed)
    tf.random.set_seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)

set_seeds(42)
print("✅ Seeds set (42)")

# ============================================================================
# STEP 2: Data Preprocessing (RGB images, binary masks)
# ============================================================================
print("\n[STEP 2] Data preprocessing...")

def resize_images_masks(img_paths, mask_paths, out_dir, size=(256,256)):
    os.makedirs(os.path.join(out_dir,"images"), exist_ok=True)
    os.makedirs(os.path.join(out_dir,"masks"), exist_ok=True)
    for ipath, mpath in tqdm(zip(img_paths, mask_paths), total=len(img_paths)):
        img = cv2.imread(ipath, cv2.IMREAD_COLOR)
        mask = cv2.imread(mpath, cv2.IMREAD_GRAYSCALE)
        img_resized = cv2.resize(img, size, interpolation=cv2.INTER_LINEAR)
        mask_resized = cv2.resize(mask, size, interpolation=cv2.INTER_NEAREST)
        mask_resized = np.where(mask_resized > 127, 255, 0).astype(np.uint8)
        fname = os.path.basename(ipath)
        cv2.imwrite(os.path.join(out_dir,"images", fname), img_resized)
        cv2.imwrite(os.path.join(out_dir,"masks", fname), mask_resized)

# Define dataset paths
base_dir = "/content/oil_spill_dataset/dataset"
train_imgs = sorted(glob.glob(os.path.join(base_dir, "train/images/*")))
train_masks = sorted(glob.glob(os.path.join(base_dir, "train/masks/*")))
val_imgs   = sorted(glob.glob(os.path.join(base_dir, "val/images/*")))
val_masks  = sorted(glob.glob(os.path.join(base_dir, "val/masks/*")))
test_imgs  = sorted(glob.glob(os.path.join(base_dir, "test/images/*")))
test_masks = sorted(glob.glob(os.path.join(base_dir, "test/masks/*")))

# Resize to 256x256
print("   Resizing images and masks to 256x256...")
resize_images_masks(train_imgs, train_masks, "/content/resized/train", size=(256,256))
resize_images_masks(val_imgs,   val_masks,   "/content/resized/val",   size=(256,256))
resize_images_masks(test_imgs,  test_masks,  "/content/resized/test",  size=(256,256))
print("✅ Data resized and saved")

# ============================================================================
# STEP 3: TensorFlow Dataset Pipeline
# ============================================================================
print("\n[STEP 3] Setting up TensorFlow dataset pipeline...")

def normalize(image, mask):
    image = tf.cast(image, tf.float32) / 255.0
    mask  = tf.cast(mask, tf.float32) / 255.0
    mask  = tf.where(mask > 0.5, 1.0, 0.0)
    return image, mask

def enhanced_augment(image, mask):
    if tf.random.uniform(()) > 0.5:
        image = tf.image.flip_left_right(image)
        mask = tf.image.flip_left_right(mask)
    if tf.random.uniform(()) > 0.5:
        image = tf.image.flip_up_down(image)
        mask = tf.image.flip_up_down(mask)
    if tf.random.uniform(()) > 0.6:
        k = tf.random.uniform([], 1, 4, dtype=tf.int32)
        image = tf.image.rot90(image, k)
        mask = tf.image.rot90(mask, k)
    if tf.random.uniform(()) > 0.6:
        scale = tf.random.uniform([], 0.85, 1.15)
        h, w = tf.shape(image)[0], tf.shape(image)[1]
        new_h = tf.cast(tf.cast(h, tf.float32)*scale, tf.int32)
        new_w = tf.cast(tf.cast(w, tf.float32)*scale, tf.int32)
        image = tf.image.resize(image, [new_h, new_w])
        mask  = tf.image.resize(mask,  [new_h, new_w], method='nearest')
        image = tf.image.resize_with_crop_or_pad(image, h, w)
        mask  = tf.image.resize_with_crop_or_pad(mask,  h, w)
    if tf.random.uniform(()) > 0.5:
        image = tf.image.random_brightness(image, 0.2)
    if tf.random.uniform(()) > 0.5:
        image = tf.image.random_contrast(image, 0.7, 1.4)
    if tf.random.uniform(()) > 0.5:
        image = tf.image.random_saturation(image, 0.7, 1.4)
    image = tf.clip_by_value(image, 0.0, 1.0)
    mask  = tf.clip_by_value(mask,  0.0, 1.0)
    return image, mask

def load_dataset_optimized(image_dir, mask_dir, augment_data=False, batch_size=8):
    image_files = sorted(glob.glob(os.path.join(image_dir, "*")))
    mask_files  = sorted(glob.glob(os.path.join(mask_dir,  "*")))

    def parse_image_mask(img_path, mask_path):
        img = tf.io.read_file(img_path)
        img = tf.image.decode_png(img, channels=3)
        mask = tf.io.read_file(mask_path)
        mask = tf.image.decode_png(mask, channels=1)
        img, mask = normalize(img, mask)
        if augment_data:
            img, mask = enhanced_augment(img, mask)
        return img, mask

    ds = tf.data.Dataset.from_tensor_slices((image_files, mask_files))
    ds = ds.map(parse_image_mask, num_parallel_calls=tf.data.AUTOTUNE)
    if augment_data:
        ds = ds.shuffle(buffer_size=2000, seed=42)
    ds = ds.batch(batch_size).prefetch(tf.data.AUTOTUNE)
    return ds

print("   Loading datasets...")
train_ds = load_dataset_optimized("/content/resized/train/images", "/content/resized/train/masks", augment_data=True,  batch_size=8)
val_ds   = load_dataset_optimized("/content/resized/val/images",   "/content/resized/val/masks",   augment_data=False, batch_size=8)
test_ds  = load_dataset_optimized("/content/resized/test/images",  "/content/resized/test/masks",  augment_data=False, batch_size=8)
print("✅ Datasets loaded (RGB 3-channel)")

# ============================================================================
# STEP 4: Define Dice Loss Helper (not a metric, for loss calculation)
# ============================================================================
def dice_loss(y_true, y_pred):
    """Dice loss function (helper, not registered)"""
    return 1 - dice_coefficient(y_true, y_pred)

def focal_loss(alpha=0.25, gamma=3.0):
    """Focal loss function (helper, not registered)"""
    def loss_fn(y_true, y_pred):
        epsilon = K.epsilon()
        y_pred = K.clip(y_pred, epsilon, 1.0 - epsilon)
        p_t = tf.where(K.equal(y_true, 1), y_pred, 1 - y_pred)
        alpha_factor = K.ones_like(y_true) * alpha
        alpha_t = tf.where(K.equal(y_true, 1), alpha_factor, 1 - alpha_factor)
        ce = -K.log(p_t)
        weight = alpha_t * K.pow(1 - p_t, gamma)
        return K.mean(weight * ce)
    return loss_fn

print("\n[STEP 4] All custom functions defined and registered ✅")

# ============================================================================
# STEP 5: Model Definition (EfficientNetB0 U-Net)
# ============================================================================
print("\n[STEP 5] Creating model architecture...")

def create_efficientnet_unet(input_shape=(256,256,3)):
    inputs = Input(input_shape)
    base_model = EfficientNetB0(weights=None, include_top=False, input_tensor=inputs)

    skip_names = [
        'block2a_expand_activation',
        'block3a_expand_activation',
        'block4a_expand_activation',
        'block6a_expand_activation',
    ]
    skips = [base_model.get_layer(name).output for name in skip_names]
    x = base_model.output

    def decoder_block(x, skip, filters):
        x = Conv2DTranspose(filters, (2,2), strides=(2,2), padding='same', kernel_initializer='he_normal')(x)
        x = Concatenate()([x, skip])
        x = Conv2D(filters, (3,3), padding='same', kernel_initializer='he_normal')(x)
        x = BatchNormalization()(x)
        x = Activation('relu')(x)
        x = Conv2D(filters, (3,3), padding='same', kernel_initializer='he_normal')(x)
        x = BatchNormalization()(x)
        x = Activation('relu')(x)
        return x

    d1 = decoder_block(x,    skips[3], 256)
    d2 = decoder_block(d1,   skips[2], 128)
    d3 = decoder_block(d2,   skips[1], 64)
    d4 = decoder_block(d3,   skips[0], 32)
    d5 = Conv2DTranspose(16, (2,2), strides=(2,2), padding='same')(d4)
    d5 = Conv2D(16, (3,3), padding='same', activation='relu')(d5)
    d5 = BatchNormalization()(d5)
    outputs = Conv2D(1, (1,1), activation='sigmoid', name='segmentation')(d5)
    model = Model(inputs, outputs)
    return model

print("   Building EfficientNetB0-U-Net architecture...")
model = create_efficientnet_unet(input_shape=(256,256,3))
print("✅ Model created")

# ============================================================================
# STEP 6: Print Model Summary
# ============================================================================
print("\n[STEP 6] Model summary:")
model.summary()

# ============================================================================
# STEP 7: Freeze Encoder (Transfer Learning Stage 1)
# ============================================================================
print("\n[STEP 7] Freezing encoder layers (transfer learning)...")
freeze_count = 0
for layer in model.layers:
    if layer.name.startswith("stem") or layer.name.startswith("block"):
        layer.trainable = False
        freeze_count += 1
print(f"✅ Froze {freeze_count} encoder layers")

# ============================================================================
# STEP 8: Compile Model (WITH ALL REGISTERED FUNCTIONS)
# ============================================================================
print("\n[STEP 8] Compiling model with registered loss and metrics...")
model.compile(
    optimizer=Adam(learning_rate=1e-3, clipnorm=1.0),
    loss=optimized_combined_loss,  # ← Registered loss
    metrics=[pixel_accuracy, dice_coefficient, iou_score]  # ← Registered metrics
)
print("✅ Model compiled")
print(f"   Total params: {model.count_params():,}")
print(f"   Trainable params: {np.sum([K.count_params(w) for w in model.trainable_weights]):,}")

# ============================================================================
# STEP 9: Define Callbacks
# ============================================================================
print("\n[STEP 9] Setting up callbacks...")
callbacks = [
    EarlyStopping(monitor='val_dice_coefficient', patience=10, mode='max', restore_best_weights=True, verbose=1),
    ModelCheckpoint('AI_SpillGuard_high_accuracy_best_v1.keras', monitor='val_dice_coefficient', save_best_only=True, mode='max', verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=5, min_lr=1e-8, verbose=1)
]
print("✅ Callbacks configured")

# ============================================================================
# STEP 10: Training Stage 1 (Encoder Frozen)
# ============================================================================
print("\n[STEP 10] STAGE 1: Training decoder with frozen encoder...")
print("   Duration: ~15 epochs")
history1 = model.fit(
    train_ds,
    epochs=15,
    validation_data=val_ds,
    callbacks=callbacks,
    verbose=1
)
print("✅ Stage 1 training complete")

# ============================================================================
# STEP 11: Training Stage 2 (Fine-tune All Layers)
# ============================================================================
print("\n[STEP 11] STAGE 2: Fine-tuning all layers...")
print("   Unfreezing all layers and reducing learning rate...")
for layer in model.layers:
    layer.trainable = True

model.compile(
    optimizer=Adam(learning_rate=1e-4, clipnorm=1.0),
    loss=optimized_combined_loss,
    metrics=[pixel_accuracy, dice_coefficient, iou_score]
)
print("   Duration: ~20 epochs")
history2 = model.fit(
    train_ds,
    epochs=20,
    validation_data=val_ds,
    callbacks=callbacks,
    verbose=1
)
print("✅ Stage 2 training complete")

# ============================================================================
# STEP 12: Evaluation on Test Set
# ============================================================================
print("\n[STEP 12] Evaluating on test set...")
model.load_weights('AI_SpillGuard_high_accuracy_best_v1.keras')
test_results = model.evaluate(test_ds, verbose=1)
metric_names = ['Loss', 'PixelAccuracy', 'DiceCoeff', 'IoU']
print("\nTest results:")
for name, val in zip(metric_names, test_results):
    print(f"  {name:>15}: {val:.4f}")

# ============================================================================
# STEP 13: Save Final Models (WITHOUT COMPILING - avoids metric issues)
# ============================================================================
print("\n[STEP 13] Saving final models with correct metadata...")

# Save as .keras (PRIMARY FORMAT)
print("   Saving as .keras format...")
model.save('AI_SpillGuard_v1.keras', save_format='keras')
size_keras = os.path.getsize('AI_SpillGuard_v1.keras') / (1024*1024)
print(f"   ✅ AI_SpillGuard_v1.keras ({size_keras:.1f} MB)")

# Save as .h5 (BACKUP FORMAT)
print("   Saving as .h5 format...")
model.save('AI_SpillGuard_v1.h5')
size_h5 = os.path.getsize('AI_SpillGuard_v1.h5') / (1024*1024)
print(f"   ✅ AI_SpillGuard_v1.h5 ({size_h5:.1f} MB)")

# Best checkpoint already saved as AI_SpillGuard_high_accuracy_best.keras
size_best = os.path.getsize('AI_SpillGuard_high_accuracy_best_v1.keras') / (1024*1024)
print(f"   ✅ AI_SpillGuard_high_accuracy_best_v1.keras ({size_best:.1f} MB)")

print("\n✅ All models saved successfully!")

# ============================================================================
# STEP 14: Verify Saved Models (WITH ALL CUSTOM OBJECTS)
# ============================================================================
print("\n[STEP 14] Verifying saved models...")

# Create custom objects dictionary
custom_objects = {
    'optimized_combined_loss': optimized_combined_loss,
    'pixel_accuracy': pixel_accuracy,
    'dice_coefficient': dice_coefficient,
    'iou_score': iou_score
}

print("\n   Test 1: Loading .keras model...")
try:
    test_model_keras = tf.keras.models.load_model(
        'AI_SpillGuard_v1.keras',
        custom_objects=custom_objects  # ← Pass ALL custom objects
    )
    print(f"   ✅ .keras model loaded successfully")
    print(f"      Input shape: {test_model_keras.input_shape}")
    print(f"      Output shape: {test_model_keras.output_shape}")

    if test_model_keras.input_shape != (None, 256, 256, 3):
        print(f"      ⚠️  WARNING: Input shape mismatch!")
    else:
        print(f"      ✅ Input shape is CORRECT: (None, 256, 256, 3)")

except Exception as e:
    print(f"   ❌ FAILED: {e}")
    import traceback
    traceback.print_exc()

print("\n   Test 2: Loading .h5 model...")
try:
    test_model_h5 = tf.keras.models.load_model(
        'AI_SpillGuard_v1.h5',
        custom_objects=custom_objects  # ← Pass ALL custom objects
    )
    print(f"   ✅ .h5 model loaded successfully")
    print(f"      Input shape: {test_model_h5.input_shape}")
    print(f"      Output shape: {test_model_h5.output_shape}")
except Exception as e:
    print(f"   ❌ FAILED: {e}")
    import traceback
    traceback.print_exc()

print("\n   Test 3: Test prediction...")
try:
    dummy_input = np.random.rand(1, 256, 256, 3).astype(np.float32)
    pred = model.predict(dummy_input, verbose=0)
    print(f"   ✅ Prediction successful")
    print(f"      Input shape: {dummy_input.shape}")
    print(f"      Output shape: {pred.shape}")
    print(f"      Output range: [{pred.min():.4f}, {pred.max():.4f}]")
except Exception as e:
    print(f"   ❌ FAILED: {e}")

# ============================================================================
# STEP 15: Prediction Function (for Deployment)
# ============================================================================
print("\n[STEP 15] Defining prediction function...")

def predict_optimized(model, image_path):
    """
    Predict segmentation mask for one image (RGB).
    Returns (original_image, probability_map, binary_mask).
    """
    img = cv2.imread(image_path, cv2.IMREAD_COLOR)
    img = cv2.resize(img, (256,256))
    img_norm = img.astype(np.float32) / 255.0
    pred = model.predict(np.expand_dims(img_norm, 0), verbose=0)[0]
    prob_map = pred.squeeze()
    bin_mask = (prob_map > 0.5).astype(np.uint8)
    return img, prob_map, bin_mask

print("✅ Prediction function defined")

# ============================================================================
# FINAL SUMMARY
# ============================================================================
print("\n" + "="*80)
print("🎉 SUCCESS! COMPLETE TRAINING & SAVING PIPELINE FINISHED")
print("="*80)

print("\n📁 FILES SAVED (Ready to download):")
print(f"   1. AI_SpillGuard_v1.keras ({size_keras:.1f} MB) ← PRIMARY")
print(f"   2. AI_SpillGuard_v1.h5 ({size_h5:.1f} MB) ← BACKUP")
print(f"   3. AI_SpillGuard_high_accuracy_best_v1.keras ({size_best:.1f} MB) ← BEST CHECKPOINT")

print("\n✅ VERIFICATION SUMMARY:")
print("   ✓ Custom loss registered before training")
print("   ✓ All metrics registered before compilation")
print("   ✓ Model trained for 35 epochs total")
print("   ✓ Both .keras and .h5 formats saved")
print("   ✓ Models loaded and verified with all custom objects")
print("   ✓ Prediction test passed")
print("   ✓ Input shape: (None, 256, 256, 3) ✓")
print("   ✓ Output shape: (None, 256, 256, 1) ✓")

print("\n📋 NEXT STEPS:")
print("   1. Download AI_SpillGuard_v1.keras from Colab Files")
print("   2. Place in: project/backend/AI_SpillGuard.keras")
print("   3. In VS Code, run: streamlit run app.py")
print("   4. Model will load without errors ✅")

print("\n💾 FOR STREAMLIT DEPLOYMENT:")
print("   Use backend/model_loader.py from provided files")
print("   It already has ALL registered custom functions")
print("   No additional configuration needed!")

print("\n" + "="*80)
print("Ready to download! Go to Files tab in Colab left sidebar →")
print("Right-click AI_SpillGuard_v1.keras → Download")
print("="*80)
```

```python
import tensorflow as tf
from tensorflow.keras.layers import *
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.applications import EfficientNetB0
import tensorflow.keras.backend as K
import numpy as np
import cv2
import glob
import os

print("="*80)
print("AI SPILLGUARD - CORRECTED COLAB TRAINING")
print("="*80)

# Register custom functions BEFORE compiling
@tf.keras.utils.register_keras_serializable(package="SpillGuard")
def dice_loss(y_true, y_pred):
    smooth = 1.0
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

@tf.keras.utils.register_keras_serializable(package="SpillGuard")
def dice_coefficient(y_true, y_pred):
    smooth = 1e-6
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

@tf.keras.utils.register_keras_serializable(package="SpillGuard")
def iou_score(y_true, y_pred):
    smooth = 1e-6
    y_pred_bin = tf.cast(y_pred > 0.5, tf.float32)
    intersection = tf.reduce_sum(y_true * y_pred_bin)
    union = tf.reduce_sum(y_true) + tf.reduce_sum(y_pred_bin) - intersection
    return (intersection + smooth) / (union + smooth)

# ============================================================================
# LOAD DATASET
# ============================================================================
print("\n[DATASET] Loading from oil_spill_dataset/dataset/...")

def load_dataset(image_dir, mask_dir, batch_size=8):
    """Load dataset from directory"""
    image_files = sorted(glob.glob(os.path.join(image_dir, "*")))
    mask_files = sorted(glob.glob(os.path.join(mask_dir, "*")))

    print(f"  Found {len(image_files)} images in {image_dir}")

    def load_image_mask(img_path, mask_path):
        # Load image
        img = tf.io.read_file(img_path)
        img = tf.image.decode_jpeg(img, channels=3)
        img = tf.image.resize(img, [256, 256])
        img = tf.cast(img, tf.float32) / 255.0

        # Load mask
        mask = tf.io.read_file(mask_path)
        mask = tf.image.decode_png(mask, channels=1)
        mask = tf.image.resize(mask, [256, 256])
        mask = tf.cast(mask, tf.float32) / 255.0
        mask = tf.where(mask > 0.5, 1.0, 0.0)

        return img, mask

    ds = tf.data.Dataset.from_tensor_slices((image_files, mask_files))
    ds = ds.map(load_image_mask, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.shuffle(buffer_size=min(1000, len(image_files)))
    ds = ds.batch(batch_size)
    ds = ds.prefetch(tf.data.AUTOTUNE)

    return ds

# Load datasets
train_ds = load_dataset('oil_spill_dataset/dataset/train/images',
                        'oil_spill_dataset/dataset/train/masks',
                        batch_size=8)
val_ds = load_dataset('oil_spill_dataset/dataset/val/images',
                      'oil_spill_dataset/dataset/val/masks',
                      batch_size=8)
test_ds = load_dataset('oil_spill_dataset/dataset/test/images',
                       'oil_spill_dataset/dataset/test/masks',
                       batch_size=8)

print("✅ Datasets loaded")

# ============================================================================
# BUILD MODEL
# ============================================================================
print("\n[MODEL] Building EfficientNetB0-UNet...")

def create_efficientnet_unet(input_shape=(256, 256, 3)):
    inputs = Input(input_shape, name="input_layer")

    # Encoder (EfficientNetB0)
    base_model = EfficientNetB0(weights=None, include_top=False, input_tensor=inputs)

    # Skip connections
    skip_names = [
        'block2a_expand_activation',
        'block3a_expand_activation',
        'block4a_expand_activation',
        'block6a_expand_activation'
    ]
    skip_outputs = [base_model.get_layer(name).output for name in skip_names]

    x = base_model.output

    # Decoder
    def decoder_block(x, skip, filters):
        x = Conv2DTranspose(filters, (2, 2), strides=(2, 2), padding='same')(x)
        x = Concatenate()([x, skip])
        x = Conv2D(filters, (3, 3), padding='same', activation='relu')(x)
        x = BatchNormalization()(x)
        x = Conv2D(filters, (3, 3), padding='same', activation='relu')(x)
        x = BatchNormalization()(x)
        return x

    d1 = decoder_block(x, skip_outputs[3], 256)
    d2 = decoder_block(d1, skip_outputs[2], 128)
    d3 = decoder_block(d2, skip_outputs[1], 64)
    d4 = decoder_block(d3, skip_outputs[0], 32)
    d5 = Conv2DTranspose(16, (2, 2), strides=(2, 2), padding='same')(d4)
    d5 = Conv2D(16, (3, 3), padding='same', activation='relu')(d5)
    d5 = BatchNormalization()(d5)

    outputs = Conv2D(1, (1, 1), activation='sigmoid', name='segmentation')(d5)

    model = Model(inputs, outputs)
    return model

model = create_efficientnet_unet()
model.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss='binary_crossentropy',
    metrics=[dice_coefficient, iou_score]
)

print("✅ Model created")
print(f"   Total params: {model.count_params():,}")

# ============================================================================
# CALLBACKS
# ============================================================================
callbacks = [
    EarlyStopping(
        monitor='val_dice_coefficient',
        patience=5,
        mode='max',
        restore_best_weights=True,
        verbose=1
    ),
    ModelCheckpoint(
        'AI_SpillGuard_v1_best.keras',
        monitor='val_dice_coefficient',
        save_best_only=True,
        mode='max',
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        min_lr=1e-7,
        verbose=1
    )
]

# ============================================================================
# TRAIN
# ============================================================================
print("\n[TRAINING] Starting training...")
print(f"Epochs: 30, Batch size: 8")

history = model.fit(
    train_ds,
    epochs=30,
    validation_data=val_ds,
    callbacks=callbacks,
    verbose=1
)

# ============================================================================
# EVALUATE
# ============================================================================
print("\n[EVALUATION] Testing on test set...")

test_results = model.evaluate(test_ds, verbose=1)
print(f"\n✅ Test Results:")
print(f"   Loss: {test_results[0]:.4f}")
print(f"   Dice Coefficient: {test_results[1]:.4f}")
print(f"   IoU Score: {test_results[2]:.4f}")

# ============================================================================
# SAVE MODEL
# ============================================================================
print("\n[SAVING] Saving model...")

# Load best checkpoint
model.load_weights('AI_SpillGuard_v1_best.keras')

# Save final model
model.save('AI_SpillGuard_v1_final_CORRECTED.keras')
print("✅ Saved: AI_SpillGuard_v1_final_CORRECTED.keras")

print("\n" + "="*80)
print("🎉 TRAINING COMPLETE!")
print("="*80)
```

```python
%%writefile backend/database.py

import sqlite3
import os
from datetime import datetime

class SpillGuardDB:
    def __init__(self, db_name="spillguard_history.db"):
        self.db_path = os.path.join(os.path.dirname(__file__), db_name)
        self.init_db()

    def init_db(self):
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS detections (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                filename TEXT NOT NULL,
                result TEXT NOT NULL,
                spill_percentage REAL NOT NULL,
                timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
            )
        ''')
        conn.commit()
        conn.close()

    def insert_detection(self, filename, result, spill_percentage):
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()
        cursor.execute('''
            INSERT INTO detections (filename, result, spill_percentage)
            VALUES (?, ?, ?)
        ''', (filename, result, spill_percentage))
        conn.commit()
        conn.close()

    def get_all_detections(self):
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()
        cursor.execute('SELECT filename, result, spill_percentage, timestamp FROM detections ORDER BY timestamp DESC')
        records = cursor.fetchall()
        conn.close()
        return records
```

```python
%%writefile backend/prediction.py
# CORRECTED prediction code for properly trained model

import numpy as np
import cv2
from PIL import Image

def preprocess_image(image_rgb, target_size=256):
    """Preprocess image for model prediction"""

    if len(image_rgb.shape) == 2:
        image_rgb = cv2.cvtColor(image_rgb, cv2.COLOR_GRAY2RGB)
    elif image_rgb.shape[2] == 4:
        image_rgb = cv2.cvtColor(image_rgb, cv2.COLOR_RGBA2RGB)

    img_resized = cv2.resize(image_rgb, (target_size, target_size), interpolation=cv2.INTER_LINEAR)

    if img_resized.dtype != np.uint8:
        img_resized = np.clip(img_resized, 0, 255).astype(np.uint8)

    img_normalized = img_resized.astype(np.float32) / 255.0

    if len(img_normalized.shape) == 2:
        img_normalized = np.stack([img_normalized] * 3, axis=-1)
    elif img_normalized.shape[2] != 3:
        if img_normalized.shape[2] == 4:
            img_normalized = img_normalized[:, :, :3]
        else:
            img_normalized = np.stack([img_normalized[:, :, 0]] * 3, axis=-1)

    img_batch = np.expand_dims(img_normalized, axis=0)

    return img_batch, img_resized


def postprocess_prediction(prediction, threshold=0.5):
    """Postprocess model prediction - for properly trained model"""

    prob_map = prediction[0, :, :, 0]
    prob_map = np.clip(prob_map, 0.0, 1.0)

    # Properly trained: HIGH (>0.5) = SPILL, LOW (<0.5) = NO SPILL
    binary_mask = (prob_map > threshold).astype(np.uint8)

    return prob_map, binary_mask


def create_overlay(image, mask, alpha=0.5):
    """Create visualization overlay"""

    if image.dtype != np.uint8:
        if image.max() <= 1.0:
            image = (image * 255).astype(np.uint8)
        else:
            image = np.clip(image, 0, 255).astype(np.uint8)

    if len(image.shape) == 2:
        image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)
    elif image.shape[2] == 4:
        image = cv2.cvtColor(image, cv2.COLOR_RGBA2RGB)

    overlay = image.copy().astype(np.float32)
    overlay[mask > 0] = [255, 0, 0]

    result = (1 - alpha) * image.astype(np.float32) + alpha * overlay
    result = np.clip(result, 0, 255).astype(np.uint8)

    return result


def calculate_spill_percentage(binary_mask):
    """Calculate spill coverage percentage"""

    total_pixels = binary_mask.size
    spill_pixels = np.sum(binary_mask)

    if total_pixels == 0:
        return 0.0

    percentage = (spill_pixels / total_pixels) * 100.0
    percentage = np.clip(percentage, 0.0, 100.0)

    return float(percentage)


def determine_risk_level(spill_percentage):
    """Determine risk level"""

    spill_percentage = float(spill_percentage)

    if spill_percentage < 1.0:
        return "✅ AREA CLEAN - No oil spill detected", "success"
    elif spill_percentage < 5.0:
        return "ℹ️ MINIMAL SPILL - Monitor area", "info"
    elif spill_percentage < 15.0:
        return "⚠️ WARNING - Moderate oil spill detected", "warning"
    else:
        return "🚨 CRITICAL - Severe oil spill detected", "danger"
```

```python
%%writefile backend/model_loader.py

import os
import warnings
warnings.filterwarnings("ignore")
import tensorflow as tf
import tensorflow.keras.backend as K
from tensorflow.keras.layers import *
from tensorflow.keras.models import Model
from tensorflow.keras.applications import EfficientNetB0

# Custom functions
def pixel_accuracy(y_true, y_pred, threshold=0.5):
    y_true_bin = K.round(y_true)
    y_pred_bin = K.cast(y_pred > threshold, dtype='float32')
    correct = K.cast(K.equal(y_true_bin, y_pred_bin), dtype='float32')
    return K.mean(correct)

def dice_coefficient(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def dice_loss(y_true, y_pred):
    return 1 - dice_coefficient(y_true, y_pred)

def focal_loss(alpha=0.25, gamma=3.0):
    def focal_loss_fixed(y_true, y_pred):
        epsilon = K.epsilon()
        y_pred = K.clip(y_pred, epsilon, 1. - epsilon)
        p_t = tf.where(K.equal(y_true, 1), y_pred, 1 - y_pred)
        alpha_factor = K.ones_like(y_true) * alpha
        alpha_t = tf.where(K.equal(y_true, 1), alpha_factor, 1 - alpha_factor)
        cross_entropy = -K.log(p_t)
        weight = alpha_t * K.pow((1 - p_t), gamma)
        loss = weight * cross_entropy
        return K.mean(loss)
    return focal_loss_fixed

def optimized_combined_loss(y_true, y_pred):
    pos_weight = tf.reduce_sum(1. - y_true) / (tf.reduce_sum(y_true) + K.epsilon())
    pos_weight = tf.clip_by_value(pos_weight, 1.0, 100.0)
    bce = K.binary_crossentropy(y_true, y_pred)
    weighted_bce = tf.where(tf.equal(y_true, 1), pos_weight * bce, bce)
    wbce_loss = K.mean(weighted_bce)
    dice = dice_loss(y_true, y_pred)
    focal = focal_loss(alpha=0.25, gamma=3.0)(y_true, y_pred)
    return 0.3 * wbce_loss + 0.5 * dice + 0.2 * focal

def iou_score(y_true, y_pred, threshold=0.5):
    y_pred_binary = tf.cast(y_pred > threshold, tf.float32)
    intersection = tf.reduce_sum(y_true * y_pred_binary)
    union = tf.reduce_sum(y_true) + tf.reduce_sum(y_pred_binary) - intersection
    return (intersection + K.epsilon()) / (union + K.epsilon())

CUSTOM_OBJECTS = {
    'pixel_accuracy': pixel_accuracy,
    'dice_coefficient': dice_coefficient,
    'dice_loss': dice_loss,
    'optimized_combined_loss': optimized_combined_loss,
    'iou_score': iou_score,
    'focal_loss': focal_loss
}

def load_spillguard_model(model_path = os.path.join(os.getcwd(), 'AI_SpillGuard_v1_final_CORRECTED.keras')
):
    if not os.path.exists(model_path):
        print(f"❌ Model not found: {model_path}")
        return None, False

    try:
        model = tf.keras.models.load_model(model_path, custom_objects=CUSTOM_OBJECTS, compile=False)
        print(f"✅ Model loaded: {model_path}")
        return model, True
    except Exception as e:
        print(f"❌ Failed: {e}")
        return None, False
```
